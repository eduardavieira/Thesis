```python
import scipy
import scipy.signal
from numpy import loadtxt
from biosppy.signals import ecg
```

# Filtering ButterWorth Filter (0.5-40Hz)

```python
def bandpassfilter(signal):
    fs=700
    lowcut=0.5
    highcut=40
    
    nyq = 0.5 * fs
    low = lowcut / nyq
    high = highcut / nyq
    
    order = 2
    
    b, a = scipy.signal.butter(order, [low,high], 'bandpass', analog=False)
    y = scipy.signal.filtfilt(b,a, signal, axis=0)
    
    return(y)

```

```python
%matplotlib qt
from matplotlib.pyplot import plot, show, title

ecg = loadtxt("S2_stress_ECG_values.csv", delimiter=',')
data = bandpassfilter(ecg)

#title("Original signal")
#plot(ecg)
#show()

title("Filtered signal")
plot(data)
show()

#from numpy import savetxt
#savetxt( "S2_stress_ECG_values_FILTERED.csv", data, delimiter=',')
```

# Detecting R Peaks with Biospy (Hamilton)

```python
from biosppy.signals import ecg
import numpy
from numpy import savetxt

filtered_ecg = data
#For Wesad
fs = 700

def rPeakHamilton(filtered_ecg, fs):
    #Returns R-peak location indices
    rPeakTuple = ecg.hamilton_segmenter(signal=filtered_ecg, sampling_rate=fs)
    
    # rPeak is classbiosppy.utils.ReturnTuple(values, names=None), must convert to ordered dictionary
    rPeakLocation = rPeakTuple.as_dict()
    rPeakLocation = rPeakLocation["rpeaks"]
    
    #Obtaining R-peak values
    rr = []
    for i in range(len(rPeakLocation)):
        xx = filtered_ecg[i]
        rr.append(xx)
        i = i + 1
    
    return rr

rPeaks = rPeakHamilton(filtered_ecg, fs)

```
